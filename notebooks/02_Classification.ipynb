```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, ConfusionMatrixDisplay
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
```

```python
seed = 0
```

```python
df = pd.read_csv('Dataset.csv',)
df
```

```python
X = df.drop(['46'], axis=1)
y = df['46']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2 , random_state=seed, stratify=y)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

# Logistic Regression

```python
lg = LogisticRegression(random_state=seed)
lg.fit(X_train, y_train)
lg_predicts = lg.predict(X_test)

print('Metrics on train set')
print(classification_report(y_train, lg.predict(X_train)))
print('-----------------------------------')
print('Metrics on test set')
print(classification_report(y_test, lg_predicts))
```

# SVM

```python
svm = SVC(kernel='rbf', random_state=seed)
svm.fit(X_train, y_train)
svm_predicts = svm.predict(X_test)

print('Metrics on train set')
print(classification_report(y_train, svm.predict(X_train)))
print('-----------------------------------')
print('Metrics on test set')
print(classification_report(y_test, svm_predicts))
```

# Random Forest

```python
rf = RandomForestClassifier(n_estimators=50, random_state=seed)
rf.fit(X_train, y_train)
rf_predicts = rf.predict(X_test)

print('Metrics on train set')
print(classification_report(y_train, rf.predict(X_train)))
print('-----------------------------------')
print('Metrics on test set')
print(classification_report(y_test, rf_predicts))
```

# KNN

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
knn_predicts = knn.predict(X_test)

print('Metrics on train set')
print(classification_report(y_train, knn.predict(X_train)))
print('-----------------------------------')
print('Metrics on test set')
print(classification_report(y_test, knn_predicts))
```

# MLP

```python
mlp = MLPClassifier(hidden_layer_sizes=(50), activation='relu', max_iter=300)
mlp.fit(X_train, y_train)
mlp_predicts = mlp.predict(X_test)

print('Metrics on train set')
print(classification_report(y_train, mlp.predict(X_train)))
print('-----------------------------------')
print('Metrics on test set')
print(classification_report(y_test, mlp_predicts))
```
